# Texas Business Date Calculator

A Python tool for calculating business days based on the Texas Comptroller of Public Accounts official holiday schedule. Perfect for legal deadlines, government filings, and business planning in Texas.

## Features

- **Calculate business days** excluding weekends and Texas state holidays
- **Manual holiday input** for precise control over observed dates
- **Excel output** with detailed date breakdowns
- **Google Colab compatible** with automatic file downloads
- **Handles holiday overlaps** when holidays fall on weekends

## Important Notes

2. **Business Day Rule Range**: Must be between 1 and 999 days

3. **Date Format**: Always use YYYY-MM-DD format (e.g., "2025-10-09")

4. **Weekend Adjustment Rules**:
   - Saturday holidays → Observed on Friday
   - Sunday holidays → Observed on Monday

````python
# Install required packages for Google Colab
try:
    import openpyxl
    import pandas as pd
except ImportError:
    print("Installing required packages...")
    import subprocess
    import sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "openpyxl", "pandas"])
    import openpyxl
    import pandas as pd
    print("Packages installed successfully!")

import datetime
from datetime import timedelta
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils.dataframe import dataframe_to_rows
import io

# Google Colab download functionality
try:
    from google.colab import files
    COLAB_ENVIRONMENT = True
    print("Google Colab environment detected - files will auto-download")
except ImportError:
    COLAB_ENVIRONMENT = False
    print("Local environment detected - files will be saved locally")

class TexasBusinessDateCalculator:
    """
    Business date calculator based on Texas state holidays.
    Allows manual input of holiday dates for Texas Comptroller of Public Accounts official holiday schedule.
    """

    def __init__(self, year=None, manual_holidays=None):
        """
        Initialize calculator with Texas state holidays for given year.

        Args:
            year: The year for calculations
            manual_holidays: Dictionary of holiday dates in format:
                {
                    'new_years': 'YYYY-MM-DD',
                    'mlk_day': 'YYYY-MM-DD',
                    'presidents_day': 'YYYY-MM-DD',
                    'texas_independence': 'YYYY-MM-DD',
                    'san_jacinto': 'YYYY-MM-DD',
                    'memorial_day': 'YYYY-MM-DD',
                    'juneteenth': 'YYYY-MM-DD',
                    'independence_day': 'YYYY-MM-DD',
                    'lbj_day': 'YYYY-MM-DD',
                    'labor_day': 'YYYY-MM-DD',
                    'columbus_day': 'YYYY-MM-DD',
                    'veterans_day': 'YYYY-MM-DD',
                    'thanksgiving': 'YYYY-MM-DD',
                    'day_after_thanksgiving': 'YYYY-MM-DD',
                    'christmas_eve': 'YYYY-MM-DD',
                    'christmas': 'YYYY-MM-DD'
                }
        """
        self.year = year or datetime.datetime.now().year
        self.manual_holidays = manual_holidays

        if manual_holidays:
            self.holidays = self._parse_manual_holidays(manual_holidays)
        else:
            self.holidays = self._get_texas_holidays(self.year)

        self.holiday_names = {}
        self._build_holiday_names()

    def _parse_manual_holidays(self, manual_holidays):
        """Parse manually input holiday dates."""
        holidays = set()

        for holiday_key, date_str in manual_holidays.items():
            if date_str:  # Only process if date is provided
                try:
                    holiday_date = datetime.datetime.strptime(date_str, '%Y-%m-%d').date()
                    holidays.add(holiday_date)
                except ValueError:
                    print(f"Warning: Invalid date format for {holiday_key}: {date_str}")

        return holidays

    def _build_holiday_names(self):
        """Build mapping of dates to holiday names."""
        if self.manual_holidays:
            # Map manual holidays to their names
            name_map = {
                'new_years': "New Year's Day",
                'mlk_day': "Martin Luther King Jr. Day",
                'presidents_day': "Presidents' Day",
                'texas_independence': "Texas Independence Day",
                'san_jacinto': "San Jacinto Day",
                'memorial_day': "Memorial Day",
                'juneteenth': "Juneteenth/Emancipation Day",
                'independence_day': "Independence Day",
                'lbj_day': "Lyndon B. Johnson Day",
                'labor_day': "Labor Day",
                'labor_day_2024': "Labor Day",
                'labor_day_2025': "Labor Day",
                'columbus_day': "Columbus Day",
                'veterans_day': "Veterans Day",
                'veterans_day_2024': "Veterans Day",
                'thanksgiving': "Thanksgiving Day",
                'thanksgiving_2024': "Thanksgiving Day",
                'day_after_thanksgiving': "Day after Thanksgiving",
                'day_after_thanksgiving_2024': "Day after Thanksgiving",
                'christmas_eve': "Christmas Eve",
                'christmas_eve_2024': "Christmas Eve",
                'christmas': "Christmas Day",
                'christmas_2024': "Christmas Day",
                'day_after_christmas_2024': "Day after Christmas"
            }

            for key, date_str in self.manual_holidays.items():
                if date_str:
                    self.holiday_names[date_str] = name_map.get(key, "Holiday")
        else:
            # Use default calculated names
            self._build_default_holiday_names()

    def _build_default_holiday_names(self):
        """Build holiday names for auto-calculated holidays."""
        self.holiday_names = {
            f"{self.year}-01-01": "New Year's Day",
            f"{self.year}-03-02": "Texas Independence Day",
            f"{self.year}-04-21": "San Jacinto Day",
            f"{self.year}-06-19": "Juneteenth",
            f"{self.year}-07-04": "Independence Day",
            f"{self.year}-08-27": "Lyndon B. Johnson Day",
            f"{self.year}-11-11": "Veterans Day",
            f"{self.year}-12-24": "Christmas Eve",
            f"{self.year}-12-25": "Christmas Day"
        }

        # Add calculated holidays
        mlk_day = self._get_nth_weekday(self.year, 1, 0, 3)
        presidents_day = self._get_nth_weekday(self.year, 2, 0, 3)
        memorial_day = self._get_last_weekday(self.year, 5, 0)
        labor_day = self._get_nth_weekday(self.year, 9, 0, 1)
        columbus_day = self._get_nth_weekday(self.year, 10, 0, 2)
        thanksgiving = self._get_nth_weekday(self.year, 11, 3, 4)

        self.holiday_names.update({
            mlk_day.strftime('%Y-%m-%d'): "Martin Luther King Jr. Day",
            presidents_day.strftime('%Y-%m-%d'): "Presidents' Day",
            memorial_day.strftime('%Y-%m-%d'): "Memorial Day",
            labor_day.strftime('%Y-%m-%d'): "Labor Day",
            columbus_day.strftime('%Y-%m-%d'): "Columbus Day",
            thanksgiving.strftime('%Y-%m-%d'): "Thanksgiving Day",
            (thanksgiving + timedelta(days=1)).strftime('%Y-%m-%d'): "Day after Thanksgiving"
        })

    def _get_texas_holidays(self, year):
        """
        Generate Texas state holidays for the given year (auto-calculation mode).
        Based on Texas Comptroller of Public Accounts official schedule.

        NOTE: This auto-calculation may not match the official schedule exactly.
        Use manual mode with official dates for accuracy.
        """
        holidays = set()

        # New Year's Day (January 1)
        new_years = datetime.date(year, 1, 1)
        holidays.add(self._adjust_holiday_for_weekend(new_years))

        # Martin Luther King Jr. Day (3rd Monday in January)
        mlk_day = self._get_nth_weekday(year, 1, 0, 3)
        holidays.add(mlk_day)

        # Presidents' Day (3rd Monday in February)
        presidents_day = self._get_nth_weekday(year, 2, 0, 3)
        holidays.add(presidents_day)

        # Memorial Day (Last Monday in May)
        memorial_day = self._get_last_weekday(year, 5, 0)
        holidays.add(memorial_day)

        # Independence Day (July 4)
        independence_day = datetime.date(year, 7, 4)
        holidays.add(self._adjust_holiday_for_weekend(independence_day))

        # Labor Day (1st Monday in September)
        labor_day = self._get_nth_weekday(year, 9, 0, 1)
        holidays.add(labor_day)

        # Veterans Day (November 11)
        veterans_day = datetime.date(year, 11, 11)
        holidays.add(self._adjust_holiday_for_weekend(veterans_day))

        # Thanksgiving Day (4th Thursday in November)
        thanksgiving = self._get_nth_weekday(year, 11, 3, 4)
        holidays.add(thanksgiving)

        # Day after Thanksgiving (Friday after Thanksgiving)
        day_after_thanksgiving = thanksgiving + timedelta(days=1)
        holidays.add(day_after_thanksgiving)

        # Christmas Eve (December 24)
        christmas_eve = datetime.date(year, 12, 24)
        # Christmas Eve observance depends on Christmas Day
        christmas = datetime.date(year, 12, 25)
        if christmas.weekday() == 3:  # If Christmas is Thursday
            # Dec 26 is also observed as shown in 2024 schedule
            holidays.add(christmas_eve)
            holidays.add(christmas)
            holidays.add(datetime.date(year, 12, 26))
        else:
            holidays.add(self._adjust_holiday_for_weekend(christmas_eve))
            holidays.add(self._adjust_holiday_for_weekend(christmas))

        # NOTE: The following are skeleton crew days (offices technically open):
        # - San Jacinto Day (April 21)
        # - Juneteenth/Emancipation Day (June 19)
        # - LBJ Day (August 27)
        # These are NOT included as they are not full office closures

        # NOTE: The following fall on weekends and are not observed:
        # - Texas Independence Day (March 2) - if Sunday
        # - Confederate Heroes Day (January 19) - if Sunday
        # - Columbus Day is not listed as an office closure

        return holidays

    def _adjust_holiday_for_weekend(self, holiday):
        """
        Adjust holiday observance for weekends.
        If holiday falls on Saturday, observe on Friday.
        If holiday falls on Sunday, observe on Monday.
        """
        if holiday.weekday() == 5:  # Saturday
            return holiday - timedelta(days=1)
        elif holiday.weekday() == 6:  # Sunday
            return holiday + timedelta(days=1)
        else:
            return holiday

    def _get_nth_weekday(self, year, month, weekday, n):
        """Get the nth occurrence of weekday in given month/year."""
        first_day = datetime.date(year, month, 1)
        days_ahead = weekday - first_day.weekday()
        if days_ahead < 0:
            days_ahead += 7
        first_occurrence = first_day + timedelta(days=days_ahead)
        return first_occurrence + timedelta(weeks=n-1)

    def _get_last_weekday(self, year, month, weekday):
        """Get the last occurrence of weekday in given month/year."""
        if month == 12:
            last_day = datetime.date(year + 1, 1, 1) - timedelta(days=1)
        else:
            last_day = datetime.date(year, month + 1, 1) - timedelta(days=1)

        days_back = (last_day.weekday() - weekday) % 7
        return last_day - timedelta(days=days_back)

    def is_business_day(self, date):
        """
        Check if a given date is a business day.
        Business day = weekday (Mon-Fri) and not a Texas state holiday.
        """
        if isinstance(date, str):
            date = datetime.datetime.strptime(date, '%Y-%m-%d').date()

        # Check if it's a weekend
        if date.weekday() >= 5:  # Saturday = 5, Sunday = 6
            return False

        # Check if it's a holiday
        return date not in self.holidays

    def get_next_business_day(self, date):
        """Get the next business day after the given date."""
        if isinstance(date, str):
            date = datetime.datetime.strptime(date, '%Y-%m-%d').date()

        next_day = date + timedelta(days=1)
        while not self.is_business_day(next_day):
            next_day += timedelta(days=1)
        return next_day

    def validate_date_format(self, date_str):
        """Validate that date string follows ISO format YYYY-MM-DD."""
        try:
            datetime.datetime.strptime(date_str, '%Y-%m-%d')
            return True
        except ValueError:
            return False

    def validate_business_day_rule(self, rule):
        """Validate business day rule is integer between 1 and 999."""
        try:
            rule_int = int(rule)
            return 1 <= rule_int <= 999
        except (ValueError, TypeError):
            return False

    def calculate_business_dates(self, notification_date, business_day_rule):
        """
        Main calculation function.

        Args:
            notification_date (str): Notification date in YYYY-MM-DD format
            business_day_rule (int): Number of business days (1-999)

        Returns:
            dict: Contains start_date, end_date, and business_days list
        """
        # Validate inputs
        if not self.validate_date_format(notification_date):
            raise ValueError("Notification date must be in YYYY-MM-DD format")

        if not self.validate_business_day_rule(business_day_rule):
            raise ValueError("Business day rule must be an integer between 1 and 999")

        # Convert inputs
        notif_date = datetime.datetime.strptime(notification_date, '%Y-%m-%d').date()
        rule = int(business_day_rule)

        # Verify notification date is a business day
        if not self.is_business_day(notif_date):
            raise ValueError(f"Notification date {notification_date} is not a business day")

        # Calculate start date (first business day after notification date)
        start_date = self.get_next_business_day(notif_date)

        # Calculate business days from start date
        business_days = []
        current_date = start_date

        for day_count in range(1, rule + 1):
            business_days.append({
                'day_number': day_count,
                'date': current_date.strftime('%Y-%m-%d'),
                'weekday': current_date.strftime('%A')
            })

            if day_count < rule:  # Don't advance after the last day
                current_date = self.get_next_business_day(current_date)

        end_date = business_days[-1]['date']

        return {
            'notification_date': notification_date,
            'start_date': start_date.strftime('%Y-%m-%d'),
            'end_date': end_date,
            'business_day_rule': rule,
            'business_days': business_days,
            'total_calendar_days': (datetime.datetime.strptime(end_date, '%Y-%m-%d').date() - start_date).days + 1
        }

    def generate_date_range_data(self, notification_date, business_day_rule):
        """
        Generate comprehensive date range data for Excel sheet 2.

        Returns:
            list: List of dictionaries with Date_Range, Business_Day_Flag, Business_Day_Count
        """
        # Get calculation results
        result = self.calculate_business_dates(notification_date, business_day_rule)

        notif_date = datetime.datetime.strptime(notification_date, '%Y-%m-%d').date()
        start_date = datetime.datetime.strptime(result['start_date'], '%Y-%m-%d').date()
        end_date = datetime.datetime.strptime(result['end_date'], '%Y-%m-%d').date()

        # Determine the full date range to include
        current_date = notif_date
        date_range_data = []
        business_day_counter = 0

        while current_date <= end_date:
            is_biz_day = self.is_business_day(current_date)

            # Determine business day count
            if current_date == notif_date and is_biz_day:
                biz_day_count = "Day 0"
            elif current_date >= start_date and is_biz_day:
                business_day_counter += 1
                biz_day_count = f"Day {business_day_counter}"
            else:
                biz_day_count = ""

            # Check if this date is a holiday
            holiday_name = ""
            if current_date in self.holidays:
                date_str = current_date.strftime('%Y-%m-%d')
                holiday_name = self.holiday_names.get(date_str, "Holiday")

            date_range_data.append({
                'Date_Range': current_date.strftime('%Y-%m-%d'),
                'Business_Day_Flag': 'Business Day' if is_biz_day else 'Non-Business Day',
                'Business_Day_Count': biz_day_count,
                'Holiday_Name': holiday_name
            })

            current_date += timedelta(days=1)

        return date_range_data

    def create_excel_workbook(self, notification_date, business_day_rule, filename=None):
        """
        Create Excel workbook with two sheets as specified.

        Args:
            notification_date (str): Notification date in YYYY-MM-DD format
            business_day_rule (int): Number of business days (1-999)
            filename (str): Optional filename, defaults to auto-generated

        Returns:
            str: Filename of created Excel file
        """
        # Get calculation results
        result = self.calculate_business_dates(notification_date, business_day_rule)

        # Generate filename if not provided
        if not filename:
            timestamp = datetime.datetime.now().strftime('%Y%m%d_%H%M%S')
            filename = f"Business_Date_Calculation_{timestamp}.xlsx"

        # Create workbook
        wb = openpyxl.Workbook()

        # Remove default sheet
        wb.remove(wb.active)

        # === SHEET 1: Summary ===
        ws1 = wb.create_sheet("Summary")

        # Header styling
        header_font = Font(bold=True, size=12)
        header_fill = PatternFill(start_color="EDEDED", end_color="EDEDED", fill_type="solid")
        center_align = Alignment(horizontal="center", vertical="center")

        # Sheet 1 data
        summary_data = [
            ["Field", "Value"],
            ["Business Day Rule", result['business_day_rule']],
            ["Notification Date", result['notification_date']],
            ["Start Date", result['start_date']],
            ["End Date", result['end_date']]
        ]

        # Write summary data
        for row_idx, row_data in enumerate(summary_data, 1):
            for col_idx, value in enumerate(row_data, 1):
                cell = ws1.cell(row=row_idx, column=col_idx, value=value)
                if row_idx == 1:  # Header row
                    cell.font = header_font
                    cell.fill = header_fill
                    cell.alignment = center_align

        # Format columns
        ws1.column_dimensions['A'].width = 20
        ws1.column_dimensions['B'].width = 15

        # === SHEET 2: Date Range ===
        ws2 = wb.create_sheet("Date_Range")

        # Generate date range data
        date_range_data = self.generate_date_range_data(notification_date, business_day_rule)

        # Create DataFrame for easier manipulation
        df = pd.DataFrame(date_range_data)

        # Write headers
        headers = ["Date_Range", "Business_Day_Flag", "Business_Day_Count", "Holiday_Name"]
        for col_idx, header in enumerate(headers, 1):
            cell = ws2.cell(row=1, column=col_idx, value=header)
            cell.font = header_font
            cell.fill = header_fill
            cell.alignment = center_align

        # Write data
        for row_idx, row in enumerate(df.itertuples(index=False), 2):
            for col_idx, value in enumerate(row, 1):
                cell = ws2.cell(row=row_idx, column=col_idx, value=value)

                # Highlight business days
                if col_idx == 2 and value == "Business Day":
                    cell.fill = PatternFill(start_color="E6F3FF", end_color="E6F3FF", fill_type="solid")

                # Highlight Day counts
                if col_idx == 3 and value and "Day" in str(value):
                    if "Day 0" in str(value):
                        cell.fill = PatternFill(start_color="FFF2CC", end_color="FFF2CC", fill_type="solid")
                    else:
                        cell.fill = PatternFill(start_color="D5E8D4", end_color="D5E8D4", fill_type="solid")

                # Highlight holidays
                if col_idx == 4 and value:
                    cell.fill = PatternFill(start_color="FFE6E6", end_color="FFE6E6", fill_type="solid")

        # Format columns
        ws2.column_dimensions['A'].width = 15
        ws2.column_dimensions['B'].width = 20
        ws2.column_dimensions['C'].width = 20
        ws2.column_dimensions['D'].width = 25

        # Save workbook
        wb.save(filename)

        # Auto-download in Google Colab
        if COLAB_ENVIRONMENT:
            print(f"Downloading {filename} to your computer...")
            files.download(filename)
        else:
            print(f"File saved as: {filename}")

        return filename

    def get_holidays_list(self):
        """Get formatted list of holidays for display."""
        sorted_holidays = []
        for holiday in sorted(self.holidays):
            date_str = holiday.strftime('%Y-%m-%d')
            name = self.holiday_names.get(date_str, "Holiday")
            sorted_holidays.append({
                'date': date_str,
                'name': name,
                'weekday': holiday.strftime('%A')
            })

        return sorted_holidays

# Example functions
def run_with_manual_holidays():
    """DEPRECATED - Use main execution block instead."""
    print("This function is deprecated. Please use the main execution block.")

def run_with_auto_holidays():
    """DEPRECATED - Use main execution block instead."""
    print("This function is deprecated. Please use the main execution block.")

# Main execution
if __name__ == "__main__":

    # ==========================================================================
    # ===================== CALCULATION PARAMETERS =============================
    # ==========================================================================

    NOTIFICATION_DATE = "2025-10-09"  # Your notification date (YYYY-MM-DD)
    BUSINESS_DAY_RULE = 60            # Number of business days (1-999)
    YEAR = 2025                       # Year for calculations

    # ==========================================================================
    # ===================== CHOOSE CALCULATION MODE ============================
    # ==========================================================================

    USE_MANUAL_MODE = True  # Set to False to use auto-calculation

    # ==========================================================================
    # ===================== MANUAL HOLIDAY DATES (if using manual mode) ========
    # ==========================================================================

    # Official Texas State Holiday Schedule from Texas Comptroller/State Auditor
    # Based on: https://comptroller.texas.gov/about/holidays.php
    # Only includes "Office Closed" holidays (not optional or skeleton crew days)
    #
    # IMPORTANT NOTES:
    # - San Jacinto Day (4/21/25): Skeleton crew required - offices technically open
    # - Emancipation Day/Juneteenth (6/19/25): Skeleton crew required - offices technically open
    # - LBJ Day (8/27/25): Skeleton crew required - offices technically open
    # - Optional holidays (Rosh Hashanah, Good Friday, Cesar Chavez Day) are not included
    # - Texas Independence Day (3/2/25) falls on Sunday - not observed as office closure
    # - Confederate Heroes Day (1/19/25) falls on Sunday - not observed as office closure

    # Define your holiday dates here in YYYY-MM-DD format
    # Leave blank ('') to skip a holiday
    MANUAL_HOLIDAYS = {
        'labor_day': '2025-09-01',                  # Labor Day
        'veterans_day': '2025-11-11',               # Veterans Day
        'day_before_thanksgiving': '2025-11-26',    # Day Before Thanksgiving
        'thanksgiving': '2025-11-27',               # Thanksgiving Day
        'day_after_thanksgiving': '2025-11-28',     # Day After Thanksgiving
        'winter_agency_holiday_1': '2025-12-22',    # Winter Agency Holiday (first day)
        'winter_agency_holiday_2': '2025-12-23',    # Winter Agency Holiday (second day)
        'christmas_eve': '2025-12-24',              # Christmas Eve
        'christmas': '2025-12-25',                  # Christmas Day
        'day_after_christmas': '2025-12-26',        # Day After Christmas
        'new_years_day': '2026-01-01',              # New Year's Day
        'martin_luther_king_jr_day': '2026-01-19',  # Martin Luther King, Jr. Day
        'presidents_day': '2026-02-16',             # Presidents' Day
        'good_friday': '2026-04-03',                # Good Friday
        'memorial_day': '2026-05-25',               # Memorial Day
        'juneteenth': '2026-06-19',                 # Juneteenth
        'summer_agency_holiday': '2026-07-03',      # Summer Agency Holiday
        'independence_day': '2026-07-04',           # Independence Day
    }

    # MANUAL_HOLIDAYS = {
        # 'labor_day_2024': '2024-09-02',      # Labor Day 2024 (if dates span into FY2025)
        # 'veterans_day_2024': '2024-11-11',   # Veterans Day 2024
        # 'thanksgiving_2024': '2024-11-28',   # Thanksgiving Day 2024
        # 'day_after_thanksgiving_2024': '2024-11-29',  # Day after Thanksgiving 2024
        # 'christmas_eve_2024': '2024-12-24',  # Christmas Eve Day 2024
        # 'christmas_2024': '2024-12-25',      # Christmas Day 2024
        # 'day_after_christmas_2024': '2024-12-26',  # Day after Christmas 2024 (observed)
        # 'new_years': '2025-01-01',           # New Year's Day
        # 'mlk_day': '2025-01-20',             # Martin Luther King Jr. Day
        # 'presidents_day': '2025-02-17',      # Presidents' Day
        # 'memorial_day': '2025-05-26',        # Memorial Day
        # 'independence_day': '2025-07-04',    # Independence Day
        # 'labor_day': '2025-09-01',           # Labor Day 2025 (estimated - 1st Monday Sept)
        # 'columbus_day': '',                  # Not listed as office closure in FY2025
        # 'veterans_day': '2025-11-10',        # Veterans Day 2025 (Nov 11 is Tuesday, likely observed Monday)
        # 'thanksgiving': '2025-11-27',        # Thanksgiving Day 2025 (estimated - 4th Thursday Nov)
        # 'day_after_thanksgiving': '2025-11-28',  # Day after Thanksgiving 2025
        # 'christmas_eve': '2025-12-24',       # Christmas Eve 2025
        # 'christmas': '2025-12-25',           # Christmas Day 2025
        # Skeleton crew days - include these ONLY if your office treats them as closures
        # Uncomment if needed:
        # 'san_jacinto': '2025-04-21',       # San Jacinto Day (Skeleton crew)
        # 'juneteenth': '2025-06-19',        # Emancipation Day/Juneteenth (Skeleton crew)
        # 'lbj_day': '2025-08-27',           # LBJ Day (Skeleton crew)
    # }

    # ==========================================================================
    # ===================== RUN CALCULATION ====================================
    # ==========================================================================

    if USE_MANUAL_MODE:
        print("=== Texas Business Date Calculator with Manual Holiday Input ===")
        print("=" * 60)
        calculator = TexasBusinessDateCalculator(manual_holidays=MANUAL_HOLIDAYS)
        mode = "Manual"
    else:
        print("=== Texas Business Date Calculator with Auto-Calculated Holidays ===")
        print("=" * 60)
        calculator = TexasBusinessDateCalculator(year=YEAR)
        mode = "Auto-Calculated"

    try:
        # Display configured holidays
        print(f"\n{mode} Texas State Holidays:")
        print("-" * 40)
        holidays = calculator.get_holidays_list()
        for h in holidays:
            print(f"{h['date']} ({h['weekday']}): {h['name']}")

        print("\n" + "=" * 60)

        # Validate notification date is a business day
        if not calculator.is_business_day(NOTIFICATION_DATE):
            print(f"ERROR: {NOTIFICATION_DATE} is not a business day!")
            print("Please choose a different notification date.")
        else:
            # Run calculation and create Excel file
            result = calculator.calculate_business_dates(NOTIFICATION_DATE, BUSINESS_DAY_RULE)
            filename = calculator.create_excel_workbook(NOTIFICATION_DATE, BUSINESS_DAY_RULE)

            print(f"\n✓ Notification Date: {NOTIFICATION_DATE}")
            print(f"✓ Business Day Rule: {BUSINESS_DAY_RULE} days")
            print(f"✓ Start Date (Day 1): {result['start_date']}")
            print(f"✓ End Date (Day {BUSINESS_DAY_RULE}): {result['end_date']}")
            print(f"✓ Total calendar days span: {result['total_calendar_days']}")
            print(f"✓ Excel file created: {filename}")

    except Exception as e:
        print(f"ERROR: {e}")
        print("\nPlease check your input values and try again.")
````

````python
# -*- coding: utf-8 -*-
"""
Texas Business Date Calculator (Manual Mode Only)
Calculates business days based on manually entered Texas state holiday dates.
"""

# Install required packages for Google Colab
try:
    import openpyxl
    import pandas as pd
except ImportError:
    print("Installing required packages...")
    import subprocess
    import sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "openpyxl", "pandas"])
    import openpyxl
    import pandas as pd
    print("Packages installed successfully!")

import datetime
from datetime import timedelta
from openpyxl.styles import Font, PatternFill, Alignment
from openpyxl.utils.dataframe import dataframe_to_rows
import io

# Google Colab download functionality
try:
    from google.colab import files
    COLAB_ENVIRONMENT = True
    print("Google Colab environment detected - files will auto-download")
except ImportError:
    COLAB_ENVIRONMENT = False
    print("Local environment detected - files will be saved locally")

class TexasBusinessDateCalculator:
    """
    Business date calculator based on manually entered Texas state holidays.
    """
    def __init__(self, manual_holidays=None):
        self.manual_holidays = manual_holidays
        self.holidays = self._parse_manual_holidays(manual_holidays)
        self.holiday_names = {}
        self._build_holiday_names()

    def _parse_manual_holidays(self, manual_holidays):
        holidays = set()
        for holiday_key, date_str in manual_holidays.items():
            if date_str:  # Only process if date is provided
                try:
                    holiday_date = datetime.datetime.strptime(date_str, '%Y-%m-%d').date()
                    holidays.add(holiday_date)
                except ValueError:
                    print(f"Warning: Invalid date format for {holiday_key}: {date_str}")
        return holidays

    def _build_holiday_names(self):
        # Map manual holidays to their names
        name_map = {
            'labor_day': "Labor Day",
            'veterans_day': "Veterans Day",
            'day_before_thanksgiving': "Day Before Thanksgiving",
            'thanksgiving': "Thanksgiving Day",
            'day_after_thanksgiving': "Day After Thanksgiving",
            'winter_agency_holiday_1': "Winter Agency Holiday (first day)",
            'winter_agency_holiday_2': "Winter Agency Holiday (second day)",
            'christmas_eve': "Christmas Eve",
            'christmas': "Christmas Day",
            'day_after_christmas': "Day After Christmas",
            'new_years_day': "New Year's Day",
            'martin_luther_king_jr_day': "Martin Luther King, Jr. Day",
            'presidents_day': "Presidents' Day",
            'good_friday': "Good Friday",
            'memorial_day': "Memorial Day",
            'juneteenth': "Juneteenth",
            'summer_agency_holiday': "Summer Agency Holiday",
            'independence_day': "Independence Day",
        }
        for key, date_str in self.manual_holidays.items():
            if date_str:
                self.holiday_names[date_str] = name_map.get(key, "Holiday")

    def is_business_day(self, date):
        if isinstance(date, str):
            date = datetime.datetime.strptime(date, '%Y-%m-%d').date()
        # Check if it's a weekend
        if date.weekday() >= 5:  # Saturday = 5, Sunday = 6
            return False
        # Check if it's a holiday
        return date not in self.holidays
    def get_next_business_day(self, date):
        if isinstance(date, str):
            date = datetime.datetime.strptime(date, '%Y-%m-%d').date()
        next_day = date + timedelta(days=1)
        while not self.is_business_day(next_day):
            next_day += timedelta(days=1)
        return next_day

    def validate_date_format(self, date_str):
        try:
            datetime.datetime.strptime(date_str, '%Y-%m-%d')
            return True
        except ValueError:
            return False

    def validate_business_day_rule(self, rule):
        try:
            rule_int = int(rule)
            return 1 <= rule_int <= 999
        except (ValueError, TypeError):
            return False

    def calculate_business_dates(self, notification_date, business_day_rule):
        # Validate inputs
        if not self.validate_date_format(notification_date):
            raise ValueError("Notification date must be in YYYY-MM-DD format")
        if not self.validate_business_day_rule(business_day_rule):
            raise ValueError("Business day rule must be an integer between 1 and 999")
        notif_date = datetime.datetime.strptime(notification_date, '%Y-%m-%d').date()
        rule = int(business_day_rule)
        # Verify notification date is a business day
        if not self.is_business_day(notif_date):
            raise ValueError(f"Notification date {notification_date} is not a business day")
        # Calculate start date (first business day after notification date)
        start_date = self.get_next_business_day(notif_date)
        # Calculate business days from start date
        business_days = []
        current_date = start_date
        for day_count in range(1, rule + 1):
            business_days.append({
                'day_number': day_count,
                'date': current_date.strftime('%Y-%m-%d'),
                'weekday': current_date.strftime('%A')
            })
            if day_count < rule:  # Don't advance after the last day
                current_date = self.get_next_business_day(current_date)
        end_date = business_days[-1]['date']
        return {
            'notification_date': notification_date,
            'start_date': start_date.strftime('%Y-%m-%d'),
            'end_date': end_date,
            'business_day_rule': rule,
            'business_days': business_days,
            'total_calendar_days': (datetime.datetime.strptime(end_date, '%Y-%m-%d').date() - start_date).days + 1
        }

    def generate_date_range_data(self, notification_date, business_day_rule):
        result = self.calculate_business_dates(notification_date, business_day_rule)
        notif_date = datetime.datetime.strptime(notification_date, '%Y-%m-%d').date()
        start_date = datetime.datetime.strptime(result['start_date'], '%Y-%m-%d').date()
        end_date = datetime.datetime.strptime(result['end_date'], '%Y-%m-%d').date()
        current_date = notif_date
        date_range_data = []
        business_day_counter = 0
        while current_date <= end_date:
            is_biz_day = self.is_business_day(current_date)
            if current_date == notif_date and is_biz_day:
                biz_day_count = "Day 0"
            elif current_date >= start_date and is_biz_day:
                business_day_counter += 1
                biz_day_count = f"Day {business_day_counter}"
            else:
                biz_day_count = ""
            holiday_name = ""
            if current_date in self.holidays:
                date_str = current_date.strftime('%Y-%m-%d')
                holiday_name = self.holiday_names.get(date_str, "Holiday")
            date_range_data.append({
                'Date_Range': current_date.strftime('%Y-%m-%d'),
                'Business_Day_Flag': 'Business Day' if is_biz_day else 'Non-Business Day',
                'Business_Day_Count': biz_day_count,
                'Holiday_Name': holiday_name
            })
            current_date += timedelta(days=1)
        return date_range_data

    def create_excel_workbook(self, notification_date, business_day_rule, filename=None):
        result = self.calculate_business_dates(notification_date, business_day_rule)
        if not filename:
            timestamp = datetime.datetime.now().strftime('%Y%m%d_%H%M%S')
            filename = f"Business_Date_Calculation_{timestamp}.xlsx"
        wb = openpyxl.Workbook()
        wb.remove(wb.active)
        # === SHEET 1: Summary ===
        ws1 = wb.create_sheet("Summary")
        header_font = Font(bold=True, size=12)
        header_fill = PatternFill(start_color="EDEDED", end_color="EDEDED", fill_type="solid")
        center_align = Alignment(horizontal="center", vertical="center")
        summary_data = [
            ["Field", "Value"],
            ["Business Day Rule", result['business_day_rule']],
            ["Notification Date", result['notification_date']],
            ["Start Date", result['start_date']],
            ["End Date", result['end_date']]
        ]
        for row_idx, row_data in enumerate(summary_data, 1):
            for col_idx, value in enumerate(row_data, 1):
                cell = ws1.cell(row=row_idx, column=col_idx, value=value)
                if row_idx == 1:
                    cell.font = header_font
                    cell.fill = header_fill
                    cell.alignment = center_align
        ws1.column_dimensions['A'].width = 20
        ws1.column_dimensions['B'].width = 15
        # === SHEET 2: Date Range ===
        ws2 = wb.create_sheet("Date_Range")
        date_range_data = self.generate_date_range_data(notification_date, business_day_rule)
        df = pd.DataFrame(date_range_data)
        headers = ["Date_Range", "Business_Day_Flag", "Business_Day_Count", "Holiday_Name"]
        for col_idx, header in enumerate(headers, 1):
            cell = ws2.cell(row=1, column=col_idx, value=header)
            cell.font = header_font
            cell.fill = header_fill
            cell.alignment = center_align
        for row_idx, row in enumerate(df.itertuples(index=False), 2):
            for col_idx, value in enumerate(row, 1):
                cell = ws2.cell(row=row_idx, column=col_idx, value=value)
                if col_idx == 2 and value == "Business Day":
                    cell.fill = PatternFill(start_color="E6F3FF", end_color="E6F3FF", fill_type="solid")
                if col_idx == 3 and value and "Day" in str(value):
                    if "Day 0" in str(value):
                        cell.fill = PatternFill(start_color="FFF2CC", end_color="FFF2CC", fill_type="solid")
                    else:
                        cell.fill = PatternFill(start_color="D5E8D4", end_color="D5E8D4", fill_type="solid")
                if col_idx == 4 and value:
                    cell.fill = PatternFill(start_color="FFE6E6", end_color="FFE6E6", fill_type="solid")
        ws2.column_dimensions['A'].width = 15
        ws2.column_dimensions['B'].width = 20
        ws2.column_dimensions['C'].width = 20
        ws2.column_dimensions['D'].width = 25
        wb.save(filename)
        if COLAB_ENVIRONMENT:
            print(f"Downloading {filename} to your computer...")
            files.download(filename)
        else:
            print(f"File saved as: {filename}")
        return filename

    def get_holidays_list(self):
        sorted_holidays = []
        for holiday in sorted(self.holidays):
            date_str = holiday.strftime('%Y-%m-%d')
            name = self.holiday_names.get(date_str, "Holiday")
            sorted_holidays.append({
                'date': date_str,
                'name': name,
                'weekday': holiday.strftime('%A')
            })
        return sorted_holidays

# ===================== CALCULATION PARAMETERS =============================

NOTIFICATION_DATE = "2025-10-09"  # Your notification date (YYYY-MM-DD)
BUSINESS_DAY_RULE = 60            # Number of business days (1-999)

# ===================== MANUAL HOLIDAY DATES ===============================
MANUAL_HOLIDAYS = {
    'labor_day': '2025-09-01',                  # Labor Day
    'veterans_day': '2025-11-11',               # Veterans Day
    'day_before_thanksgiving': '2025-11-26',    # Day Before Thanksgiving
    'thanksgiving': '2025-11-27',               # Thanksgiving Day
    'day_after_thanksgiving': '2025-11-28',     # Day After Thanksgiving
    'winter_agency_holiday_1': '2025-12-22',    # Winter Agency Holiday (first day)
    'winter_agency_holiday_2': '2025-12-23',    # Winter Agency Holiday (second day)
    'christmas_eve': '2025-12-24',              # Christmas Eve
    'christmas': '2025-12-25',                  # Christmas Day
    'day_after_christmas': '2025-12-26',        # Day After Christmas
    'new_years_day': '2026-01-01',              # New Year's Day
    'martin_luther_king_jr_day': '2026-01-19',  # Martin Luther King, Jr. Day
    'presidents_day': '2026-02-16',             # Presidents' Day
    'good_friday': '2026-04-03',                # Good Friday
    'memorial_day': '2026-05-25',               # Memorial Day
    'juneteenth': '2026-06-19',                 # Juneteenth
    'summer_agency_holiday': '2026-07-03',      # Summer Agency Holiday
    'independence_day': '2026-07-04',           # Independence Day
}

if __name__ == "__main__":
    print("=== Texas Business Date Calculator with Manual Holiday Input ===")
    print("=" * 60)
    calculator = TexasBusinessDateCalculator(manual_holidays=MANUAL_HOLIDAYS)
    try:
        # Display configured holidays
        print("\nManual Texas State Holidays:")
        print("-" * 40)
        holidays = calculator.get_holidays_list()
        for h in holidays:
            print(f"{h['date']} ({h['weekday']}): {h['name']}")
        print("\n" + "=" * 60)
        # Validate notification date is a business day
        if not calculator.is_business_day(NOTIFICATION_DATE):
            print(f"ERROR: {NOTIFICATION_DATE} is not a business day!")
            print("Please choose a different notification date.")
        else:
            # Run calculation and create Excel file
            result = calculator.calculate_business_dates(NOTIFICATION_DATE, BUSINESS_DAY_RULE)
            filename = calculator.create_excel_workbook(NOTIFICATION_DATE, BUSINESS_DAY_RULE)
            print(f"\n✓ Notification Date: {NOTIFICATION_DATE}")
            print(f"✓ Business Day Rule: {BUSINESS_DAY_RULE} days")
            print(f"✓ Start Date (Day 1): {result['start_date']}")
            print(f"✓ End Date (Day {BUSINESS_DAY_RULE}): {result['end_date']}")
            print(f"✓ Total calendar days span: {result['total_calendar_days']}")
            print(f"✓ Excel file created: {filename}")
    except Exception as e:
        print(f"ERROR: {e}")
        print("\nPlease check your input values and try again.")
````

````python
# -*- coding: utf-8 -*-
"""
Texas Business Date Calculator (Manual Mode Only)
Calculates business days based on manually entered Texas state holiday dates.

Output:
- Excel workbook with:
  - Sheet "Summary": parameters and results
  - Sheet "Date_Range": full date span with flags and NonBusiness_Reason
"""

# Install required packages for Google Colab (safe no-op on local if installed)
try:
    import openpyxl
    import pandas as pd
except ImportError:
    print("Installing required packages...")
    import subprocess
    import sys
    subprocess.check_call([sys.executable, "-m", "pip", "install", "openpyxl", "pandas"])
    import openpyxl
    import pandas as pd
    print("Packages installed successfully!")

import datetime
from datetime import timedelta
from openpyxl.styles import Font, PatternFill, Alignment

# Google Colab download functionality
try:
    from google.colab import files
    COLAB_ENVIRONMENT = True
    print("Google Colab environment detected - files will auto-download")
except ImportError:
    COLAB_ENVIRONMENT = False
    print("Local environment detected - files will be saved locally")


class TexasBusinessDateCalculator:
    """
    Business date calculator using manually provided Texas state holiday dates only.
    """

    def __init__(self, manual_holidays=None):
        """
        Args:
            manual_holidays (dict): { 'holiday_key': 'YYYY-MM-DD', ... }
                                    Leave '' (empty string) to skip a holiday.
        """
        self.manual_holidays = manual_holidays or {}
        self.holidays = self._parse_manual_holidays(self.manual_holidays)  # set[date]
        self.holiday_names = {}  # dict[date_str] -> display name
        self._build_holiday_names()

    def _parse_manual_holidays(self, manual_holidays):
        """Parse manually input holiday dates into a set of date objects."""
        holidays = set()
        for holiday_key, date_str in manual_holidays.items():
            if date_str:  # Only process if date is provided
                try:
                    holiday_date = datetime.datetime.strptime(date_str, '%Y-%m-%d').date()
                    holidays.add(holiday_date)
                except ValueError:
                    print(f"Warning: Invalid date format for {holiday_key}: {date_str}")
        return holidays

    def _build_holiday_names(self):
        """Build mapping of ISO date strings to holiday display names."""
        name_map = {
            'labor_day': "Labor Day",
            'veterans_day': "Veterans Day",
            'day_before_thanksgiving': "Day Before Thanksgiving",
            'thanksgiving': "Thanksgiving Day",
            'day_after_thanksgiving': "Day After Thanksgiving",
            'winter_agency_holiday_1': "Winter Agency Holiday (first day)",
            'winter_agency_holiday_2': "Winter Agency Holiday (second day)",
            'christmas_eve': "Christmas Eve",
            'christmas': "Christmas Day",
            'day_after_christmas': "Day After Christmas",
            'new_years_day': "New Year's Day",
            'martin_luther_king_jr_day': "Martin Luther King, Jr. Day",
            'presidents_day': "Presidents' Day",
            'good_friday': "Good Friday",
            'memorial_day': "Memorial Day",
            'juneteenth': "Juneteenth",
            'summer_agency_holiday': "Summer Agency Holiday",
            'independence_day': "Independence Day",
        }
        for key, date_str in self.manual_holidays.items():
            if date_str:
                # If two keys ever point to the same date, the last one wins in this map,
                # but the per-row logic will still aggregate all names for that date.
                self.holiday_names[date_str] = name_map.get(key, "Holiday")

    # ------------------------- Core Business-Day Logic -------------------------

    def is_business_day(self, date):
        """
        Business day = weekday (Mon-Fri) AND not in manual holiday set.
        Accepts str 'YYYY-MM-DD' or datetime.date.
        """
        if isinstance(date, str):
            date = datetime.datetime.strptime(date, '%Y-%m-%d').date()
        # Weekend?
        if date.weekday() >= 5:  # Saturday = 5, Sunday = 6
            return False
        # Holiday?
        return date not in self.holidays

    def get_next_business_day(self, date):
        """Get the next business day strictly after the given date."""
        if isinstance(date, str):
            date = datetime.datetime.strptime(date, '%Y-%m-%d').date()
        next_day = date + timedelta(days=1)
        while not self.is_business_day(next_day):
            next_day += timedelta(days=1)
        return next_day

    # --------------------------- Validation Helpers ---------------------------

    @staticmethod
    def validate_date_format(date_str):
        """Validate that date string follows ISO format YYYY-MM-DD."""
        try:
            datetime.datetime.strptime(date_str, '%Y-%m-%d')
            return True
        except ValueError:
            return False

    @staticmethod
    def validate_business_day_rule(rule):
        """Validate business day rule is integer between 1 and 999."""
        try:
            rule_int = int(rule)
            return 1 <= rule_int <= 999
        except (ValueError, TypeError):
            return False

    # --------------------------- Calculation Methods --------------------------

    def calculate_business_dates(self, notification_date, business_day_rule):
        """
        Calculate start date (Day 1) and end date (Day N) counting only business days.
        Raises ValueError if inputs invalid or notification date is not a business day.
        """
        # Validate inputs
        if not self.validate_date_format(notification_date):
            raise ValueError("Notification date must be in YYYY-MM-DD format")
        if not self.validate_business_day_rule(business_day_rule):
            raise ValueError("Business day rule must be an integer between 1 and 999")

        notif_date = datetime.datetime.strptime(notification_date, '%Y-%m-%d').date()
        rule = int(business_day_rule)

        # Verify notification date is a business day
        if not self.is_business_day(notif_date):
            raise ValueError(f"Notification date {notification_date} is not a business day")

        # Calculate start date (first business day after notification date)
        start_date = self.get_next_business_day(notif_date)

        # Calculate business days from start date
        business_days = []
        current_date = start_date
        for day_count in range(1, rule + 1):
            business_days.append({
                'day_number': day_count,
                'date': current_date.strftime('%Y-%m-%d'),
                'weekday': current_date.strftime('%A')
            })
            if day_count < rule:  # Don't advance after the last day
                current_date = self.get_next_business_day(current_date)

        end_date = business_days[-1]['date']
        return {
            'notification_date': notification_date,
            'start_date': start_date.strftime('%Y-%m-%d'),
            'end_date': end_date,
            'business_day_rule': rule,
            'business_days': business_days,
            'total_calendar_days': (datetime.datetime.strptime(end_date, '%Y-%m-%d').date() - start_date).days + 1
        }

    def generate_date_range_data(self, notification_date, business_day_rule):
        """
        Build the detailed date span rows for Excel with:
          - Date_Range
          - Business_Day_Flag
          - Business_Day_Count
          - NonBusiness_Reason (holiday names or 'Weekend' or '')
        """
        result = self.calculate_business_dates(notification_date, business_day_rule)
        notif_date = datetime.datetime.strptime(notification_date, '%Y-%m-%d').date()
        start_date = datetime.datetime.strptime(result['start_date'], '%Y-%m-%d').date()
        end_date = datetime.datetime.strptime(result['end_date'], '%Y-%m-%d').date()

        current_date = notif_date
        date_range_data = []
        business_day_counter = 0

        while current_date <= end_date:
            is_biz_day = self.is_business_day(current_date)

            # Day counter display
            if current_date == notif_date and is_biz_day:
                biz_day_count = "Day 0"
            elif current_date >= start_date and is_biz_day:
                business_day_counter += 1
                biz_day_count = f"Day {business_day_counter}"
            else:
                biz_day_count = ""

            # ---- Determine NonBusiness_Reason ----
            nonbusiness_reason = ""
            date_str = current_date.strftime('%Y-%m-%d')

            # Aggregate all holiday names for this date (if multiple keys map to the same date)
            holiday_names = []
            for key, h_date in self.manual_holidays.items():
                if h_date == date_str:
                    holiday_names.append(self.holiday_names.get(h_date, "Holiday"))

            if holiday_names:
                nonbusiness_reason = ", ".join(sorted(set(holiday_names)))
            elif current_date.weekday() >= 5:  # Saturday or Sunday
                nonbusiness_reason = "Weekend"
            # --------------------------------------

            date_range_data.append({
                'Date_Range': date_str,
                'Business_Day_Flag': 'Business Day' if is_biz_day else 'Non-Business Day',
                'Business_Day_Count': biz_day_count,
                'NonBusiness_Reason': nonbusiness_reason
            })

            current_date += timedelta(days=1)

        return date_range_data

    # ------------------------------ Excel Output ------------------------------

    def create_excel_workbook(self, notification_date, business_day_rule, filename=None):
        """Create the Excel workbook with Summary and Date_Range sheets."""
        result = self.calculate_business_dates(notification_date, business_day_rule)

        if not filename:
            timestamp = datetime.datetime.now().strftime('%Y%m%d_%H%M%S')
            filename = f"Business_Date_Calculation_{timestamp}.xlsx"

        wb = openpyxl.Workbook()
        wb.remove(wb.active)

        # Reusable styles
        header_font = Font(bold=True, size=12)
        header_fill = PatternFill(start_color="EDEDED", end_color="EDEDED", fill_type="solid")
        center_align = Alignment(horizontal="center", vertical="center")

        fill_business_day = PatternFill(start_color="E6F3FF", end_color="E6F3FF", fill_type="solid")  # light blue
        fill_day0 = PatternFill(start_color="FFF2CC", end_color="FFF2CC", fill_type="solid")          # yellow
        fill_dayN = PatternFill(start_color="D5E8D4", end_color="D5E8D4", fill_type="solid")          # green
        fill_holiday = PatternFill(start_color="FFE6E6", end_color="FFE6E6", fill_type="solid")       # light red
        fill_weekend = PatternFill(start_color="F0F0F0", end_color="F0F0F0", fill_type="solid")       # light gray

        # === SHEET 1: Summary ===
        ws1 = wb.create_sheet("Summary")
        summary_data = [
            ["Field", "Value"],
            ["Business Day Rule", result['business_day_rule']],
            ["Notification Date", result['notification_date']],
            ["Start Date", result['start_date']],
            ["End Date", result['end_date']]
        ]
        for row_idx, row_data in enumerate(summary_data, 1):
            for col_idx, value in enumerate(row_data, 1):
                cell = ws1.cell(row=row_idx, column=col_idx, value=value)
                if row_idx == 1:  # header
                    cell.font = header_font
                    cell.fill = header_fill
                    cell.alignment = center_align

        ws1.column_dimensions['A'].width = 20
        ws1.column_dimensions['B'].width = 15

        # === SHEET 2: Date Range ===
        ws2 = wb.create_sheet("Date_Range")

        # Generate date range data
        date_range_data = self.generate_date_range_data(notification_date, business_day_rule)
        df = pd.DataFrame(date_range_data)

        # Headers
        headers = ["Date_Range", "Business_Day_Flag", "Business_Day_Count", "NonBusiness_Reason"]
        for col_idx, header in enumerate(headers, 1):
            cell = ws2.cell(row=1, column=col_idx, value=header)
            cell.font = header_font
            cell.fill = header_fill
            cell.alignment = center_align

        # Rows
        for row_idx, row in enumerate(df.itertuples(index=False), 2):
            for col_idx, value in enumerate(row, 1):
                cell = ws2.cell(row=row_idx, column=col_idx, value=value)

                # Highlight business days (light blue)
                if col_idx == 2 and value == "Business Day":
                    cell.fill = fill_business_day

                # Highlight Day counts (yellow for Day 0, green otherwise)
                if col_idx == 3 and value and "Day" in str(value):
                    if "Day 0" in str(value):
                        cell.fill = fill_day0
                    else:
                        cell.fill = fill_dayN

                # Distinct highlighting for NonBusiness_Reason:
                # - "Weekend" => light gray
                # - Holiday names => light red
                if col_idx == 4 and value:
                    if str(value).strip().lower() == "weekend":
                        cell.fill = fill_weekend
                    else:
                        cell.fill = fill_holiday

        # Column widths
        ws2.column_dimensions['A'].width = 15
        ws2.column_dimensions['B'].width = 20
        ws2.column_dimensions['C'].width = 20
        ws2.column_dimensions['D'].width = 28

        wb.save(filename)
        if COLAB_ENVIRONMENT:
            print(f"Downloading {filename} to your computer...")
            files.download(filename)
        else:
            print(f"File saved as: {filename}")
        return filename

    # ---------------------------- Convenience API -----------------------------

    def get_holidays_list(self):
        """Return a list of configured holidays: [{date, name, weekday}, ...]."""
        sorted_list = []
        for holiday in sorted(self.holidays):
            date_str = holiday.strftime('%Y-%m-%d')
            name = self.holiday_names.get(date_str, "Holiday")
            sorted_list.append({
                'date': date_str,
                'name': name,
                'weekday': holiday.strftime('%A')
            })
        return sorted_list


# ===================== CALCULATION PARAMETERS =============================

NOTIFICATION_DATE = "2025-10-09"  # Your notification date (YYYY-MM-DD)
BUSINESS_DAY_RULE = 60            # Number of business days (1-999)

# ===================== MANUAL HOLIDAY DATES (FY26 TEA) ====================

# Leave blank ('') to skip a holiday
MANUAL_HOLIDAYS = {
    'labor_day': '2025-09-01',                  # Labor Day
    'veterans_day': '2025-11-11',               # Veterans Day
    'day_before_thanksgiving': '2025-11-26',    # Day Before Thanksgiving
    'thanksgiving': '2025-11-27',               # Thanksgiving Day
    'day_after_thanksgiving': '2025-11-28',     # Day After Thanksgiving
    'winter_agency_holiday_1': '2025-12-22',    # Winter Agency Holiday (first day)
    'winter_agency_holiday_2': '2025-12-23',    # Winter Agency Holiday (second day)
    'christmas_eve': '2025-12-24',              # Christmas Eve
    'christmas': '2025-12-25',                  # Christmas Day
    'day_after_christmas': '2025-12-26',        # Day After Christmas
    'new_years_day': '2026-01-01',              # New Year's Day
    'martin_luther_king_jr_day': '2026-01-19',  # Martin Luther King, Jr. Day
    'presidents_day': '2026-02-16',             # Presidents' Day
    'good_friday': '2026-04-03',                # Good Friday
    'memorial_day': '2026-05-25',               # Memorial Day
    'juneteenth': '2026-06-19',                 # Juneteenth
    'summer_agency_holiday': '2026-07-03',      # Summer Agency Holiday
    'independence_day': '2026-07-04',           # Independence Day
}

# ===================== MAIN EXECUTION =====================================

if __name__ == "__main__":
    print("=== Texas Business Date Calculator with Manual Holiday Input ===")
    print("=" * 60)
    calculator = TexasBusinessDateCalculator(manual_holidays=MANUAL_HOLIDAYS)
    try:
        # Display configured holidays
        print("\nManual Texas State Holidays:")
        print("-" * 40)
        holidays = calculator.get_holidays_list()
        for h in holidays:
            print(f"{h['date']} ({h['weekday']}): {h['name']}")
        print("\n" + "=" * 60)

        # Validate notification date is a business day
        if not calculator.is_business_day(NOTIFICATION_DATE):
            print(f"ERROR: {NOTIFICATION_DATE} is not a business day!")
            print("Please choose a different notification date.")
        else:
            # Run calculation and create Excel file
            result = calculator.calculate_business_dates(NOTIFICATION_DATE, BUSINESS_DAY_RULE)
            filename = calculator.create_excel_workbook(NOTIFICATION_DATE, BUSINESS_DAY_RULE)
            print(f"\n✓ Notification Date: {NOTIFICATION_DATE}")
            print(f"✓ Business Day Rule: {BUSINESS_DAY_RULE} days")
            print(f"✓ Start Date (Day 1): {result['start_date']}")
            print(f"✓ End Date (Day {BUSINESS_DAY_RULE}): {result['end_date']}")
            print(f"✓ Total calendar days span: {result['total_calendar_days']}")
            print(f"✓ Excel file created: {filename}")
    except Exception as e:
        print(f"ERROR: {e}")
        print("\nPlease check your input values and try again.")
````
